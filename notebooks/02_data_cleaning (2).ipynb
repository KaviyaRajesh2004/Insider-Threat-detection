# 01 – Data Understanding

**Goal:** Load the employee activity CSV and understand its structure before changing any data.

> This project uses a simulated learning dataset. Its labels and patterns are synthetic, not evidence of real employee misconduct.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = PROJECT_ROOT / "dataset"
MODEL_DIR = PROJECT_ROOT / "models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)

import pandas as pd
import numpy as np

data_path = DATA_DIR / "employee_activity.csv"
df = pd.read_csv(data_path)

print("Dataset loaded successfully")
print("Shape:", df.shape)
df.head()
```

```python
# Show the last five rows
df.tail()
```

```python
# Check column names and data types
df.info()
```

```python
# Statistical summary of numeric columns
df.describe().T
```

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Check target label distribution
df["threat_label"].value_counts().rename(index={0: "Normal", 1: "Suspicious"})
```

```python
# Check the percentage of each class
(df["threat_label"].value_counts(normalize=True) * 100).round(2)
```
