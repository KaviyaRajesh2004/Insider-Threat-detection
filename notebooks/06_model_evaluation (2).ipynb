# 03 – Exploratory Data Analysis (EDA)

**Goal:** Explore activity patterns and compare the simulated Normal and Suspicious groups visually.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = PROJECT_ROOT / "dataset"
MODEL_DIR = PROJECT_ROOT / "models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(DATA_DIR / "employee_activity_cleaned.csv")
sns.set_theme()
df.head()
```

```python
# Target class distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="threat_label")
plt.title("Normal vs Suspicious Records")
plt.xlabel("Threat label (0 = Normal, 1 = Suspicious)")
plt.ylabel("Record count")
plt.tight_layout()
plt.show()
```

```python
# Distribution of downloaded files by label
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="threat_label", y="files_downloaded")
plt.title("Files Downloaded by Threat Label")
plt.xlabel("Threat label")
plt.ylabel("Files downloaded")
plt.tight_layout()
plt.show()
```

```python
# Compare average activity by label
activity_cols = [
    "login_after_hours",
    "files_downloaded",
    "usb_connected",
    "failed_logins",
    "sensitive_file_access"
]
df.groupby("threat_label")[activity_cols].mean().round(2)
```

```python
# Correlation heatmap for numeric columns
plt.figure(figsize=(9, 6))
sns.heatmap(df.select_dtypes(include="number").corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Count of after-hours logins by threat label
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="login_after_hours", hue="threat_label")
plt.title("After-hours Login vs Threat Label")
plt.xlabel("Login after hours (0 = No, 1 = Yes)")
plt.ylabel("Record count")
plt.tight_layout()
plt.show()
```
